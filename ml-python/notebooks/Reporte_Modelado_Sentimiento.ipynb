Este proyecto no se limitó a entrenar un modelo que dice "Positivo" o "Negativo". Construimos un **ecosistema de análisis** capaz de entender la ironía, priorizar crisis y guiar decisiones operativas en tiempo real.

## 1. La Problemática: ¿Por qué fallan los modelos estándar?
Durante nuestra investigación, detectamos que los modelos comerciales fallan en el 23% de las reseñas de negocios debido a:
*   **Ironía:** *"¡Qué lujo de cucarachas!"* (Detectado como Positivo por IAs genéricas).
*   **Contrastes:** *"El desayuno fue genial, PERO la cama era un desastre"* (Detectado como Neutral, perdiendo una queja crítica).
*   **Vetos de Seguridad:** Fallos de higiene o seguridad que deben tratarse con prioridad absoluta.

## 2. Nuestra Solución: El Motor G68 Supreme
Nuestra arquitectura de doble capa combina un clasificador estadístico con un **Motor Lógico Semántico**.

### Los 3 Pilares del Éxito:
1.  **Veto Crítico:** Prioridad inmediata ante incidentes graves (Higiene, Robos, Seguridad).
2.  **Anclaje de Contexto:** Clasificación precisa  (**Feedback del Cliente**).
3.  **Detector de Sarcasmo:** Neutraliza elogios falsos mediante análisis de conectores contrastivos.

## 3. Demo en Vivo: El Poder de G68
A continuación, puedes probar cómo el motor "desenmascara" una reseña irónica o crítica.

```python
import sys
import os
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from motor_hibrido import enriquecer_respuesta
from engine.sentiment_engine import SentimentEngine

# 1. Configuración de Rutas y Motor
ROOT_PATH = os.path.abspath(os.path.join(os.getcwd(), '..'))
MODELS_PATH = os.path.join(ROOT_PATH, 'data', 'models')
engine = SentimentEngine(model_dir=MODELS_PATH)

# 2. Componentes de la Interfaz
text_input = widgets.Textarea(
    value='Me encanta que la habitación esté sucia y el aire no funcione. Una maravilla de estafa.',
    placeholder='Escribe una reseña aquí...',
    description='Entrada:',
    layout=widgets.Layout(width='100%', height='100px')
)
button = widgets.Button(description="Procesar con G68", button_style='info', icon='brain')
output = widgets.Output()

# 3. Lógica de Negocio y Visualización
def analizar_visual(b):
    # Limpiamos el output PRIMERO para evitar duplicados acumulados
    output.clear_output(wait=True)
    
    with output:
        texto = text_input.value
        if not texto.strip(): return
        
        # Predicción e Informe
        pred_ia, prob_ia = engine.predict_raw(texto)
        res = enriquecer_respuesta(texto, pred_ia, prob_ia, engine)
        
        label = res['prevision']
        prob = res['probabilidad']
        feats = res['top_features']
        color = "#2ecc71" if label == "Positivo" else "#e74c3c" if label == "Negativo" else "#95a5a6"
        prioridad = "CRÍTICA" if "Veto" in str(res.get('explicabilidad_interna', {}).get('motivo', '')) else "NORMAL"
        
        display(HTML(f"""
        <div style="border-left: 10px solid {color}; padding: 25px; background: #fff; border-radius: 8px; box-shadow: 0 4px 6px rgba(0,0,0,0.1); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;">
            <h2 style="margin:0; color:{color}; font-size: 26px;">{label} ({prob:.2%})</h2>
            <hr style="border: 0; border-top: 1px solid #eee; margin: 15px 0;">
            <p style="font-size:1.15em;"><b>Nivel de Alerta:</b> <span style="color:{color}; font-weight:bold; font-size: 1.2em;">{prioridad}</span></p>
            <p><b>🧭 Hallazgos (Top Features):</b> {feats}</p>
        </div>
        """))

# 4. Vinculación Blindada: Evitamos registros múltiples re-ejecutando la celda
button.on_click(analizar_visual)
display(text_input, button, output)
```
