## 3. Playground Interactivo G68
Prueba el motor con frases complejas o irónicas.

```python
import sys
import os
import joblib
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Ajuste de rutas para importar el motor híbrido
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..', 'src', 'app')))
from motor_hibrido import enriquecer_respuesta

# Carga de modelos desde ml-python/data/models
model_dir = os.path.abspath(os.path.join(os.getcwd(), '..', 'data', 'models'))
model = joblib.load(os.path.join(model_dir, 'sentiment_model.pkl'))
vectorizer = joblib.load(os.path.join(model_dir, 'tfidf_vectorizer.pkl'))

text_input = widgets.Textarea(
    value='Me encanta que la habitación esté sucia y el aire no funcione. Brillantísimo.',
    placeholder='Escribe una reseña irónica aquí...',
    description='Entrada:',
    layout=widgets.Layout(width='100%', height='100px')
)
button = widgets.Button(description="Procesar G68", button_style='success', icon='brain')
output = widgets.Output()

def procesar(b):
    with output:
        clear_output()
        texto = text_input.value
        if not texto.strip(): return
        
        # Predicción IA base
        vec = vectorizer.transform([texto])
        pred_ia = model.predict(vec)[0]
        prob_ia = np.max(model.predict_proba(vec))
        
        # Enriquecimiento Motor Híbrido
        res = enriquecer_respuesta(texto, pred_ia, prob_ia)
        
        color = res['explicabilidad']['visualizacion_frontend']['color_alerta']
        prioridad = res['explicabilidad']['visualizacion_frontend']['prioridad']
        
        display(HTML(f"""
        <div style="border-left: 10px solid {color}; padding: 20px; background: #fdfdfd; border-radius: 5px; box-shadow: 2px 2px 5px #eee;">
            <h2 style="margin:0; color:{color};">{res['previsión']} ({res['probabilidad']:.2%})</h2>
            <p style="font-size:1.2em; border-bottom: 1px solid #ddd; padding-bottom:10px;"><b>Prioridad:</b> {prioridad}</p>
            <p><b>🔍 Hallazgos:</b> {', '.join(res['explicabilidad']['hallazgos'])}</p>
            <p><b>🏢 Clasificación:</b> {', '.join(res['explicabilidad']['departamentos'])}</p>
        </div>
        """))

import numpy as np
button.on_click(procesar)
display(text_input, button, output)
```
